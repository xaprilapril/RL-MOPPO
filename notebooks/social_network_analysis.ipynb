## 1. Carga del grafo y estadísticas básicas

Usamos el **ego-network de Facebook** (SNAP, Leskovec & Mcauley 2012).  
Se extrae un subgrafo de 100 nodos mediante BFS desde el hub de mayor grado.

```python
G = load_facebook_ego(n_nodes=100, seed=42, data_dir=DATA_DIR)
n, m = G.number_of_nodes(), G.number_of_edges()

density        = nx.density(G)
avg_clustering = nx.average_clustering(G)
transitivity   = nx.transitivity(G)
avg_degree     = 2 * m / n
degree_seq     = sorted([d for _, d in G.degree()], reverse=True)
max_degree     = max(degree_seq)
min_degree     = min(degree_seq)

is_connected   = nx.is_connected(G)
n_components   = nx.number_connected_components(G)
giant          = max(nx.connected_components(G), key=len)
G_giant        = G.subgraph(giant).copy()
diameter       = nx.diameter(G_giant)
avg_path_len   = nx.average_shortest_path_length(G_giant)

stats_dict = {
    'Nodos (n)'              : n,
    'Aristas (m)'            : m,
    'Densidad'               : round(density, 5),
    'Grado promedio'         : round(avg_degree, 3),
    'Grado máximo'           : max_degree,
    'Grado mínimo'           : min_degree,
    'Clustering promedio'    : round(avg_clustering, 5),
    'Transitividad (C_delta)': round(transitivity, 5),
    'Diámetro'               : diameter,
    'Camino promedio'        : round(avg_path_len, 4),
    'Componentes conexas'    : n_components,
    'Tamaño comp. gigante'   : len(giant),
}

df_stats = pd.DataFrame.from_dict(stats_dict, orient='index', columns=['Valor'])
print('=== Propiedades globales de la red ego-Facebook (n=100) ===\n')
print(df_stats.to_string())
```

## 2. Distribución de grados

Las redes sociales reales siguen distribuciones de grado **heavy-tail** (aproximación a ley de potencia P(k) ~ k^{-gamma}).  
Estimamos gamma con regresión log-log y calculamos la asimetría (skewness).

```python
degrees     = [d for _, d in G.degree()]
deg_counter = Counter(degrees)
unique_degs = sorted(deg_counter.keys())
freqs       = [deg_counter[d] for d in unique_degs]

log_k = np.log(unique_degs)
log_p = np.log([f / n for f in freqs])
slope, intercept, r_val, p_val, _ = stats.linregress(log_k, log_p)
gamma = -slope

deg_array = np.array(degrees)
print(f'gamma estimado (ley de potencia): {gamma:.3f}  (R2={r_val**2:.3f})')
print(f'Skewness : {stats.skew(deg_array):.3f}  (>0 = cola derecha = hubs)')
print(f'Kurtosis : {stats.kurtosis(deg_array):.3f}')
print(f'Clases de grado unicas: {len(unique_degs)}')

singleton_degrees = [d for d, cnt in deg_counter.items() if cnt == 1]
singleton_nodes   = [nd for nd, d in G.degree() if d in singleton_degrees]
print(f'\nNodos con grado unico (no 2-anonimos): {len(singleton_nodes)}/{n}')
print(f'  r_priv inicial = {compute_k_anonymity_reward(G, k=2):.5f}')

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].bar(unique_degs, freqs, color='#1E88E5', alpha=0.8, edgecolor='white')
axes[0].set_xlabel('Grado k'); axes[0].set_ylabel('Frecuencia')
axes[0].set_title('Distribucion de grados')

k_fit = np.linspace(min(unique_degs), max(unique_degs), 100)
p_fit = np.exp(intercept) * k_fit ** slope
axes[1].scatter(unique_degs, [f/n for f in freqs], color='#1E88E5', s=50, zorder=3, label='Datos')
axes[1].plot(k_fit, p_fit, 'r--', linewidth=1.5, label=f'Ajuste gamma={gamma:.2f}')
axes[1].set_xscale('log'); axes[1].set_yscale('log')
axes[1].set_xlabel('log k'); axes[1].set_ylabel('log P(k)')
axes[1].set_title(f'Log-log (R2={r_val**2:.3f})')
axes[1].legend(fontsize=8)

sorted_deg = np.sort(deg_array)
ccdf = 1 - np.arange(1, n + 1) / n
axes[2].step(sorted_deg, ccdf, color='#43A047', linewidth=2)
axes[2].set_xlabel('Grado k'); axes[2].set_ylabel('P(K > k)')
axes[2].set_title('CCDF (cola de distribucion)')
axes[2].set_xscale('log'); axes[2].set_yscale('log')

plt.suptitle('Analisis de distribucion de grados — ego-Facebook (n=100)', fontsize=12)
plt.tight_layout()
plt.savefig(f'{OUT_DIR}/sna_degree_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
print('Guardado: sna_degree_distribution.png')
```

## 3. Medidas de centralidad

La **centralidad** mide la importancia de un nodo en la red:

| Metrica | Definicion | Interpretacion |
|---------|-----------|----------------|
| **Degree** | k_v / (n-1) | conexiones directas |
| **Betweenness** | fraccion de caminos minimos que pasan por v | cuantos flujos controla |
| **Closeness** | 1 / avg_dist(v, todos) | que tan rapido alcanza a todos |
| **Eigenvector** | importancia ponderada por vecinos importantes | ser amigo de gente importante |
| **PageRank** | variante de eigenvector con damping | version usada por Google |

```python
print('Calculando centralidades...')
cent_degree    = nx.degree_centrality(G)
cent_between   = nx.betweenness_centrality(G, normalized=True)
cent_closeness = nx.closeness_centrality(G)
cent_eigenvec  = nx.eigenvector_centrality(G, max_iter=1000)
cent_pagerank  = nx.pagerank(G, alpha=0.85)

df_cent = pd.DataFrame({
    'Degree'      : cent_degree,
    'Betweenness' : cent_between,
    'Closeness'   : cent_closeness,
    'Eigenvector' : cent_eigenvec,
    'PageRank'    : cent_pagerank,
})
df_cent['Degree_raw'] = [G.degree(nd) for nd in df_cent.index]

print('\nTop-10 nodos por Betweenness Centrality:')
top10 = df_cent.sort_values('Betweenness', ascending=False).head(10)
print(top10[['Degree_raw', 'Betweenness', 'Closeness', 'Eigenvector', 'PageRank']].to_string())

print('\nCorrelaciones entre centralidades:')
print(df_cent[['Degree','Betweenness','Closeness','Eigenvector','PageRank']].corr().round(3).to_string())

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sc = axes[0].scatter(df_cent['Degree'], df_cent['Betweenness'],
                     c=df_cent['PageRank'], cmap='plasma', s=60, alpha=0.8)
plt.colorbar(sc, ax=axes[0], label='PageRank')
axes[0].set_xlabel('Degree centrality')
axes[0].set_ylabel('Betweenness centrality')
axes[0].set_title('Degree vs. Betweenness\n(color = PageRank)')

metrics  = ['Degree', 'Betweenness', 'Closeness', 'Eigenvector']
colors_c = ['#1E88E5', '#E53935', '#43A047', '#FB8C00']
x = np.arange(5)
w = 0.2
for i, (metric, color) in enumerate(zip(metrics, colors_c)):
    top5 = df_cent.nlargest(5, metric)
    vals = top5[metric].values
    axes[1].bar(x + i * w, vals / vals.max(), w, label=metric, color=color, alpha=0.85)
axes[1].set_xticks(x + 1.5 * w)
axes[1].set_xticklabels([f'#{i+1}' for i in range(5)])
axes[1].set_ylabel('Centralidad normalizada (sobre su maximo)')
axes[1].set_title('Top-5 nodos por cada centralidad')
axes[1].legend(fontsize=8)

plt.suptitle('Centralidades — ego-Facebook (n=100)', fontsize=12)
plt.tight_layout()
plt.savefig(f'{OUT_DIR}/sna_centrality.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 4. Deteccion de comunidades

Una **comunidad** es un subconjunto de nodos con mas conexiones internas que externas.  
Usamos **Greedy Modularity** (Clauset et al. 2004), que maximiza:

Q = (1/2m) * sum_{i,j} [ A_{ij} - k_i*k_j/(2m) ] * delta(c_i, c_j)

Q > 0.3 indica estructura comunitaria significativa.

```python
try:
    import community as community_louvain
    partition = community_louvain.best_partition(G, random_state=42)
    method = 'Louvain'
except ImportError:
    communities_gen = nx.community.greedy_modularity_communities(G)
    partition = {}
    for cid, comm in enumerate(communities_gen):
        for node in comm:
            partition[node] = cid
    method = 'Greedy Modularity'

n_communities = len(set(partition.values()))
modularity    = nx.community.modularity(
    G,
    [{nd for nd, c in partition.items() if c == cid} for cid in set(partition.values())]
)

comm_sizes = Counter(partition.values())
print(f'Metodo       : {method}')
print(f'Comunidades  : {n_communities}')
print(f'Modularidad Q: {modularity:.4f}  ({"Estructura significativa" if modularity > 0.3 else "Estructura debil"})')
print(f'Tamanos      : {sorted(comm_sizes.values(), reverse=True)}')

rows = []
for cid in sorted(comm_sizes.keys()):
    members = [nd for nd, c in partition.items() if c == cid]
    subg = G.subgraph(members)
    rows.append({
        'Comunidad'        : cid,
        'Tamano'           : len(members),
        'Aristas internas' : subg.number_of_edges(),
        'Densidad interna' : round(nx.density(subg), 4) if len(members) > 1 else 0,
        'Clustering prom.' : round(nx.average_clustering(subg), 4) if len(members) > 1 else 0,
    })
df_comm = pd.DataFrame(rows).set_index('Comunidad')
print('\nResumen por comunidad:')
print(df_comm.to_string())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

palette    = list(mcolors.TABLEAU_COLORS.values())
node_colors = [palette[partition[nd] % len(palette)] for nd in G.nodes()]
node_sizes  = [30 + 5 * G.degree(nd) for nd in G.nodes()]
pos         = nx.spring_layout(G, seed=42, k=0.4)

nx.draw_networkx_edges(G, pos, ax=axes[0], alpha=0.2, edge_color='gray', width=0.5)
nx.draw_networkx_nodes(G, pos, ax=axes[0],
                       node_color=node_colors, node_size=node_sizes, alpha=0.9)
top5_nodes = df_cent.nlargest(5, 'Degree').index.tolist()
nx.draw_networkx_labels(G, pos, {nd: str(nd) for nd in top5_nodes},
                        ax=axes[0], font_size=7, font_weight='bold')
axes[0].set_title(f'Red ego-Facebook — {n_communities} comunidades ({method})\n'
                  f'n={n}, m={m}, Q={modularity:.3f}')
axes[0].axis('off')

comm_ids = sorted(set(partition.values()))
inter_mat = np.zeros((len(comm_ids), len(comm_ids)))
for u, v in G.edges():
    cu, cv = partition[u], partition[v]
    inter_mat[cu][cv] += 1
    if cu != cv:
        inter_mat[cv][cu] += 1

im = axes[1].imshow(inter_mat, cmap='Blues', aspect='auto')
axes[1].set_xticks(range(len(comm_ids)))
axes[1].set_yticks(range(len(comm_ids)))
axes[1].set_xticklabels([f'C{c}' for c in comm_ids])
axes[1].set_yticklabels([f'C{c}' for c in comm_ids])
axes[1].set_title('Matriz de conexiones inter-comunidades')
plt.colorbar(im, ax=axes[1], label='N aristas')
for i in range(len(comm_ids)):
    for j in range(len(comm_ids)):
        axes[1].text(j, i, int(inter_mat[i, j]), ha='center', va='center',
                     fontsize=8,
                     color='white' if inter_mat[i, j] > inter_mat.max() * 0.6 else 'black')

plt.tight_layout()
plt.savefig(f'{OUT_DIR}/sna_communities.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 5. Coeficiente de clustering

El **clustering local** de un nodo v mide que fraccion de sus vecinos estan tambien conectados entre si:

C_v = 2 * |{e_jk}| / (k_v * (k_v - 1))

**Fenomeno conocido en redes sociales:** C(k) ~ k^{-beta}  
Nodos perifericos (bajo grado) tienen clustering alto; hubs conectan comunidades distintas (clustering bajo).

```python
clust_local = nx.clustering(G)

deg_to_clust = {}
for node in G.nodes():
    d = G.degree(node)
    deg_to_clust.setdefault(d, []).append(clust_local[node])
ck_x = sorted(deg_to_clust.keys())
ck_y = [np.mean(deg_to_clust[d]) for d in ck_x]

valid = [(x, y) for x, y in zip(ck_x, ck_y) if x > 0 and y > 0]
if valid:
    lx, ly    = zip(*valid)
    b_slope, b_int, *_ = stats.linregress(np.log(lx), np.log(ly))
    beta      = -b_slope
else:
    beta      = float('nan')
    b_slope, b_int = 0, 0

print(f'Clustering global C: {avg_clustering:.5f}')
print(f'Transitividad      : {transitivity:.5f}')
print(f'Exponente C(k)     : beta = {beta:.3f}')
print('\nEstadisticas del clustering local:')
print(pd.Series(list(clust_local.values())).describe().round(4).to_string())

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].scatter(list(clust_local.keys()), list(clust_local.values()),
                c=[G.degree(nd) for nd in clust_local], cmap='viridis', s=50, alpha=0.7)
axes[0].set_xlabel('Nodo'); axes[0].set_ylabel('Clustering local C_v')
axes[0].set_title('Clustering local por nodo (color = grado)')

axes[1].scatter(ck_x, ck_y, color='#1E88E5', s=60, zorder=3, label='C(k) observado')
if not np.isnan(beta) and len(valid) > 1:
    k_fit  = np.linspace(min(ck_x) + 0.1, max(ck_x), 100)
    ck_fit = np.exp(b_int) * k_fit ** b_slope
    axes[1].plot(k_fit, ck_fit, 'r--', label=f'C(k)~k^(-{beta:.2f})')
axes[1].set_xscale('log'); axes[1].set_yscale('log')
axes[1].set_xlabel('Grado k (log)'); axes[1].set_ylabel('C(k) promedio (log)')
axes[1].set_title('Clustering vs. grado C(k)')
axes[1].legend(fontsize=8)

plt.suptitle('Analisis de clustering — ego-Facebook', fontsize=12)
plt.tight_layout()
plt.savefig(f'{OUT_DIR}/sna_clustering.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 6. Analisis de caminos y fenomeno small-world

Las redes sociales exhiben el **"small world"** (Watts & Strogatz, 1998):
- Distancia media L corta (logaritmica en n)
- Clustering alto (mucho mayor que grafo aleatorio equivalente)

Indice de mundo pequeno: sigma = (C/C_rand) / (L/L_rand) > 1

```python
p_rand = 2 * m / (n * (n - 1))
G_rand = nx.gnp_random_graph(n, p_rand, seed=42)
gc_r   = max(nx.connected_components(G_rand), key=len)
G_rand = G_rand.subgraph(gc_r).copy()
G_rand = nx.relabel_nodes(G_rand, {old: i for i, old in enumerate(G_rand.nodes())})

L_real = avg_path_len
C_real = avg_clustering
L_rand = nx.average_shortest_path_length(G_rand)
C_rand = nx.average_clustering(G_rand)
sigma  = (C_real / C_rand) / (L_real / L_rand)

all_dists = []
for source in G_giant.nodes():
    sp = nx.single_source_shortest_path_length(G_giant, source)
    all_dists.extend(v for v in sp.values() if v > 0)

print('=== Fenomeno Small World ===')
print(f'{"Metrica":<30} {"Red real":>12} {"Erdos-Renyi":>12}')
print('-' * 56)
print(f'{"Camino promedio L":<30} {L_real:>12.4f} {L_rand:>12.4f}')
print(f'{"Clustering C":<30} {C_real:>12.4f} {C_rand:>12.4f}')
print(f'{"Diametro":<30} {diameter:>12} {nx.diameter(G_rand):>12}')
print(f'{"C/C_rand":<30} {C_real/C_rand:>12.3f} {1.0:>12.3f}')
print(f'{"L/L_rand":<30} {L_real/L_rand:>12.3f} {1.0:>12.3f}')
print(f'{"sigma (small-world index)":<30} {sigma:>12.3f} {1.0:>12.3f}')
print(f'\nResultado: sigma={sigma:.2f} > 1 -> {"RED SMALL-WORLD" if sigma > 1 else "No small-world"}')

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].hist(all_dists, bins=range(1, diameter + 2), color='#1E88E5',
             alpha=0.8, edgecolor='white', rwidth=0.85)
axes[0].axvline(L_real, color='red', linestyle='--', label=f'L={L_real:.2f}')
axes[0].set_xlabel('Distancia geodesica'); axes[0].set_ylabel('Frecuencia')
axes[0].set_title('Distribucion de distancias entre todos los pares')
axes[0].legend()

cats      = ['Clustering C', 'Camino L (norm.)']
vals_real = [C_real, L_real / max(L_real, L_rand)]
vals_rand = [C_rand, L_rand / max(L_real, L_rand)]
xb = np.arange(len(cats)); wb = 0.35
axes[1].bar(xb - wb/2, vals_real, wb, label='ego-Facebook', color='#1E88E5')
axes[1].bar(xb + wb/2, vals_rand, wb, label='Erdos-Renyi', color='#90A4AE')
axes[1].set_xticks(xb); axes[1].set_xticklabels(cats)
axes[1].set_title(f'Small-world: sigma={sigma:.2f}\nAlto C + bajo L vs. aleatorio')
axes[1].legend()

plt.suptitle('Analisis de caminos — ego-Facebook', fontsize=12)
plt.tight_layout()
plt.savefig(f'{OUT_DIR}/sna_paths.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 8. Impacto de la anonimizacion k-grado sobre la estructura de red

Aqui conectamos ambas materias: aplicamos **k-degree anonymization** y medimos como cambian las metricas SNA.

**Pregunta empirica:** La anonimizacion destruye la estructura social de la red?

Tres versiones del grafo:
- **G_orig**: grafo original
- **G_greedy**: anonimizado con Greedy (Liu & Terzi 2008) — solo agrega aristas
- **G_random**: modificado aleatoriamente (cota inferior, referencia)

```python
def greedy_k_anonymize(G0, k=2, max_ops=3000):
    G_anon = G0.copy()
    for _ in range(max_ops):
        deg_counts = {}
        for _, d in G_anon.degree():
            deg_counts[d] = deg_counts.get(d, 0) + 1
        at_risk = [nd for nd, d in G_anon.degree() if deg_counts[d] < k]
        if not at_risk:
            break
        at_risk.sort(key=lambda nd: G_anon.degree(nd))
        made_change = False
        for u in at_risk:
            u_deg = G_anon.degree(u)
            same  = [v for v, d in G_anon.degree()
                     if d == u_deg and v != u and not G_anon.has_edge(u, v)]
            if same:
                G_anon.add_edge(u, same[0]); made_change = True; break
            other = sorted([v for v in G_anon.nodes() if v != u and not G_anon.has_edge(u, v)],
                           key=lambda v: -deg_counts.get(G_anon.degree(v), 0))
            if other:
                G_anon.add_edge(u, other[0]); made_change = True; break
        if not made_change:
            break
    return G_anon

def random_modify(G0, T=50, seed=0):
    rng  = np.random.default_rng(seed)
    G_r  = G0.copy()
    nodes = list(G_r.nodes())
    all_e = list(combinations(nodes, 2))
    for _ in range(T):
        u, v = all_e[rng.integers(len(all_e))]
        G_r.remove_edge(u, v) if G_r.has_edge(u, v) else G_r.add_edge(u, v)
    return G_r

print('Generando versiones anonimizadas...')
G_greedy = greedy_k_anonymize(G, k=2)
G_rnd    = random_modify(G, T=50)

def net_metrics(Gx, ref, label):
    gc = Gx.subgraph(max(nx.connected_components(Gx), key=len)).copy()
    return {
        'label'          : label,
        'n_edges'        : Gx.number_of_edges(),
        'density'        : round(nx.density(Gx), 5),
        'avg_degree'     : round(2 * Gx.number_of_edges() / Gx.number_of_nodes(), 3),
        'avg_clustering' : round(nx.average_clustering(Gx), 5),
        'transitivity'   : round(nx.transitivity(Gx), 5),
        'diameter'       : nx.diameter(gc),
        'avg_path'       : round(nx.average_shortest_path_length(gc), 4),
        'modularity'     : round(nx.community.modularity(
            Gx,
            [{nd for nd, c in partition.items() if c == cid}
             for cid in set(partition.values())]
        ), 4),
        'r_priv'         : round(compute_k_anonymity_reward(Gx, k=2), 5),
        'r_util'         : round(compute_utility_reward(Gx, ref), 5),
    }

results_impact = [
    net_metrics(G,        G, 'Original'),
    net_metrics(G_greedy, G, 'Greedy (k=2)'),
    net_metrics(G_rnd,    G, 'Aleatorio (T=50)'),
]
df_impact = pd.DataFrame(results_impact).set_index('label')

# Verificar k-anonimato del Greedy
dc_g = Counter(d for _, d in G_greedy.degree())
min_cls = min(dc_g.values())
print(f'\nGreedy: aristas anadidas={G_greedy.number_of_edges()-m}, '
      f'min_clase={min_cls}, k-anonimo={min_cls >= 2}')

print('\nComparacion de metricas:')
print(df_impact.T.to_string())
```
