## 1. Carga de Datasets

Comparamos todos los presets. `facebook` descarga `facebook_combined.txt.gz` de SNAP (~900 KB) y lo cachea en `data/`.

```python
from graph_anon_morl.datasets import load_facebook_ego

print('Cargando datasets...')
graphs = {
    'karate'       : nx.karate_club_graph(),
    'barbell-20'   : nx.barbell_graph(10, 1),
    'random-20'    : nx.erdos_renyi_graph(20, 0.3, seed=42),
    'facebook-50'  : load_facebook_ego(n_nodes=50,  seed=42, data_dir=DATA_DIR),
    'facebook-100' : load_facebook_ego(n_nodes=100, seed=42, data_dir=DATA_DIR),
}
print('Listo.\n')

rows = []
for name, G in graphs.items():
    n = G.number_of_nodes()
    rows.append({
        'preset'         : name,
        'nodes'          : n,
        'edges'          : G.number_of_edges(),
        'action_space'   : n * (n - 1) // 2,
        'density'        : round(nx.density(G), 4),
        'avg_degree'     : round(sum(d for _, d in G.degree()) / n, 2),
        'avg_clustering' : round(nx.average_clustering(G), 4),
        'is_connected'   : nx.is_connected(G),
    })

df_presets = pd.DataFrame(rows).set_index('preset')
display(df_presets)
```

## 2. EDA — ego-Facebook (n=100)

Dataset principal del paper. Subgrafo BFS de 100 nodos desde el nodo de mayor grado del ego-Facebook de SNAP (4039 nodos, 88234 aristas totales).

```python
# --- Estadisticas basicas ---
G_fb = graphs['facebook-100']
n    = G_fb.number_of_nodes()
deg_vals = [d for _, d in G_fb.degree()]

is_conn = nx.is_connected(G_fb)
stats = [
    ('Nodos',                    n),
    ('Aristas',                  G_fb.number_of_edges()),
    ('Action space n*(n-1)/2',   n * (n - 1) // 2),
    ('Densidad',                 round(nx.density(G_fb), 6)),
    ('Grado promedio',           round(np.mean(deg_vals), 4)),
    ('Grado maximo',             max(deg_vals)),
    ('Grado minimo',             min(deg_vals)),
    ('Desv std grado',           round(np.std(deg_vals), 4)),
    ('Clustering promedio',      round(nx.average_clustering(G_fb), 6)),
    ('Transitividad global',     round(nx.transitivity(G_fb), 6)),
    ('Es conexo',                is_conn),
    ('Componentes conexas',      nx.number_connected_components(G_fb)),
    ('Diametro',                 nx.diameter(G_fb) if is_conn else 'N/A'),
    ('Radio',                    nx.radius(G_fb)   if is_conn else 'N/A'),
    ('Asortatividad (grado)',     round(nx.degree_assortativity_coefficient(G_fb), 6)),
]

print('ego-Facebook (n=100) — Estadisticas basicas')
print('=' * 52)
for key, val in stats:
    print(f'{key:35s}: {val}')
```

```python
# --- Distribucion de grados ---
deg_vals = sorted([d for _, d in G_fb.degree()])
deg_series = pd.Series(deg_vals)
deg_count  = deg_series.value_counts().sort_index()

fig, axes = plt.subplots(1, 3, figsize=(16, 4))

# Histograma
axes[0].hist(deg_vals, bins=20, color='steelblue', edgecolor='white')
axes[0].axvline(np.mean(deg_vals), color='red', linestyle='--',
                label=f'media={np.mean(deg_vals):.1f}')
axes[0].set_xlabel('Grado')
axes[0].set_ylabel('Frecuencia')
axes[0].set_title('Histograma de grados')
axes[0].legend()

# Log-log (chequeo de ley de potencia)
axes[1].loglog(deg_count.index, deg_count.values, 'o', color='steelblue', markersize=6)
axes[1].set_xlabel('Grado (log)')
axes[1].set_ylabel('Frecuencia (log)')
axes[1].set_title('Distribucion log-log')

# CDF empirica
axes[2].plot(np.sort(deg_vals), np.linspace(0, 1, len(deg_vals)),
             color='steelblue', linewidth=2)
axes[2].set_xlabel('Grado')
axes[2].set_ylabel('CDF')
axes[2].set_title('CDF empirica del grado')

plt.suptitle('ego-Facebook (n=100) — Analisis de grados', y=1.02, fontsize=13)
plt.tight_layout()
plt.show()

print(f'Media: {deg_series.mean():.2f}  |  Mediana: {deg_series.median():.1f}  |  '
      f'Std: {deg_series.std():.2f}  |  Max: {deg_series.max()}  |  Min: {deg_series.min()}')
```

```python
# --- Analisis de k-Anonimato por secuencia de grados (linea de base del paper) ---
degree_to_nodes = {}
for node, d in G_fb.degree():
    degree_to_nodes.setdefault(d, []).append(node)

class_sizes = sorted([len(v) for v in degree_to_nodes.values()])
min_k       = min(class_sizes)
n_singleton = sum(1 for s in class_sizes if s == 1)
n_at_risk   = sum(len(v) for v in degree_to_nodes.values() if len(v) < 2)

print('k-Anonimato por clases de equivalencia de grado (ANTES de anonimizar)')
print('=' * 60)
print(f'Clases de equivalencia (degrees unicos)  : {len(degree_to_nodes)}')
print(f'k natural (min clase)                    : {min_k}')
print(f'Clases singleton (k=1, re-identificable) : {n_singleton}')
print(f'Nodos en clases singleton                : {n_at_risk} de {G_fb.number_of_nodes()}')
print()

size_dist = pd.Series(class_sizes).value_counts().sort_index()
df_k = pd.DataFrame({
    'class_size': size_dist.index,
    'n_classes' : size_dist.values,
    'n_nodes'   : size_dist.index * size_dist.values,
    'at_risk'   : size_dist.index < 2,
})
display(df_k.set_index('class_size'))

# Visualizacion
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

colors = ['#E53935' if s < 2 else '#1E88E5' for s in class_sizes]
axes[0].bar(range(len(class_sizes)), class_sizes,
            color=colors, edgecolor='white', width=0.85)
axes[0].axhline(y=2, color='red', linestyle='--', linewidth=1.5, label='k=2 umbral')
axes[0].set_xlabel('Clase de equivalencia (ordenada por tamano)')
axes[0].set_ylabel('Tamano de clase (nodos con mismo grado)')
axes[0].set_title('Clases de equivalencia — ANTES de anonimizar\nRojo = singleton (totalmente re-identificable)')
axes[0].legend()

risk_map = {d: (len(v) < 2) for d, v in degree_to_nodes.items()}
node_sorted = sorted(G_fb.degree(), key=lambda x: x[1])
sorted_degs = [d for _, d in node_sorted]
bar_colors  = ['#E53935' if risk_map[d] else '#1E88E5' for d in sorted_degs]
axes[1].bar(range(len(sorted_degs)), sorted_degs, color=bar_colors, edgecolor='none')
axes[1].set_xlabel('Nodos (ordenados por grado)')
axes[1].set_ylabel('Grado')
axes[1].set_title('Grado por nodo  |  Rojo = en clase singleton')

plt.tight_layout()
plt.show()
```

```python
# --- Clustering coefficient ---
clustering = dict(nx.clustering(G_fb))
clust_vals = [clustering[n] for n in G_fb.nodes()]
deg_list   = [G_fb.degree(n) for n in G_fb.nodes()]

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].hist(clust_vals, bins=20, color='#43A047', edgecolor='white')
axes[0].axvline(np.mean(clust_vals), color='red', linestyle='--',
                label=f'media={np.mean(clust_vals):.3f}')
axes[0].set_xlabel('Clustering coefficient local')
axes[0].set_ylabel('Frecuencia')
axes[0].set_title('Distribucion del clustering coefficient')
axes[0].legend()

axes[1].scatter(deg_list, clust_vals, alpha=0.6, color='#43A047', s=40)
axes[1].set_xlabel('Grado')
axes[1].set_ylabel('Clustering coefficient')
axes[1].set_title('Grado vs. clustering coefficient')

plt.suptitle('ego-Facebook (n=100) — Clustering', y=1.02)
plt.tight_layout()
plt.show()

print(f'Clustering promedio  : {np.mean(clust_vals):.6f}')
print(f'Transitividad global : {nx.transitivity(G_fb):.6f}')
print(f'Diferencia           : {abs(np.mean(clust_vals) - nx.transitivity(G_fb)):.6f}')
```

```python
# --- Visualizacion del grafo ---
from matplotlib.patches import Patch

pos          = nx.spring_layout(G_fb, seed=42)
node_degrees = np.array([G_fb.degree(nd) for nd in G_fb.nodes()])
q25 = np.percentile(node_degrees, 25)
q50 = np.percentile(node_degrees, 50)
q75 = np.percentile(node_degrees, 75)

fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# Panel 1: por grado continuo
nx.draw_networkx_edges(G_fb, pos, ax=axes[0], alpha=0.2, edge_color='gray', width=0.5)
sc = nx.draw_networkx_nodes(G_fb, pos, ax=axes[0],
                             node_color=node_degrees, cmap='YlOrRd',
                             node_size=90, alpha=0.9)
plt.colorbar(sc, ax=axes[0], label='Grado')
axes[0].set_title('ego-Facebook (n=100) — coloreado por grado')
axes[0].axis('off')

# Panel 2: por cuartil (coincide con el audit de equidad)
q_colors = [
    '#2196F3' if d <= q25 else
    '#4CAF50' if d <= q50 else
    '#FF9800' if d <= q75 else
    '#F44336'
    for d in node_degrees
]
nx.draw_networkx_edges(G_fb, pos, ax=axes[1], alpha=0.2, edge_color='gray', width=0.5)
nx.draw_networkx_nodes(G_fb, pos, ax=axes[1],
                        node_color=q_colors, node_size=90, alpha=0.9)
axes[1].set_title('Por cuartil de grado  (Q1 periferia <-> Q4 hub)')
axes[1].axis('off')
legend_elems = [
    Patch(facecolor='#2196F3', label=f'Q1 periferia (deg <= {q25:.0f})'),
    Patch(facecolor='#4CAF50', label=f'Q2 (deg <= {q50:.0f})'),
    Patch(facecolor='#FF9800', label=f'Q3 (deg <= {q75:.0f})'),
    Patch(facecolor='#F44336', label=f'Q4 hub (deg > {q75:.0f})'),
]
axes[1].legend(handles=legend_elems, loc='upper right', fontsize=8)

plt.tight_layout()
plt.show()
```

```python
# --- Pre-analisis de cuartiles (espejo del audit module) ---
degrees_dict = dict(G_fb.degree())
vals_list    = list(degrees_dict.values())
q1_t = np.percentile(vals_list, 25)
q3_t = np.percentile(vals_list, 75)

def quartile_label(d):
    if d <= q1_t: return 'Q1_peripheral'
    if d >= q3_t: return 'Q4_hub'
    return 'Q2Q3_middle'

groups = {}
for node, d in degrees_dict.items():
    groups.setdefault(quartile_label(d), []).append(node)

print('Desglose por cuartil de grado (pre-anonimizacion):')
print(f'  Umbral Q1 (<=) : {q1_t:.1f}')
print(f'  Umbral Q4 (>=) : {q3_t:.1f}')
print()

rows_q = []
for lbl in ('Q1_peripheral', 'Q2Q3_middle', 'Q4_hub'):
    nodes = groups.get(lbl, [])
    if not nodes:
        continue
    node_degs = [degrees_dict[nd] for nd in nodes]
    rows_q.append({
        'grupo'     : lbl,
        'n_nodos'   : len(nodes),
        'deg_min'   : min(node_degs),
        'deg_max'   : max(node_degs),
        'deg_media' : round(np.mean(node_degs), 2),
        'deg_std'   : round(np.std(node_degs), 2),
    })

display(pd.DataFrame(rows_q).set_index('grupo'))
```

## 3. Test `GraphAnonEnv`

Verifica: observation/action spaces, reset, step, reward vector, rollout aleatorio.

```python
from graph_anon_morl.env import GraphAnonEnv

env = GraphAnonEnv(G_fb, k=2, T=50)
print('GraphAnonEnv inicializado:')
print(f'  obs_dim   : {env.observation_space.shape[0]}  (n^2 + 2 = {env.n}^2 + 2)')
print(f'  n_actions : {env.n_actions}  (n*(n-1)//2 pares no dirigidos)')
print(f'  T (steps) : {env.T}')
print(f'  k-anon k  : {env.k}')

# Reset con peso explicito
obs, _ = env.reset(weight=[0.7, 0.3])
print(f'\nReset weight=[0.7, 0.3]:')
print(f'  obs.shape : {obs.shape}')
print(f'  weight    : {env.weight}')
print(f'  suma adj  : {obs[:env.n*env.n].sum():.0f}  (= 2 * {G_fb.number_of_edges()} aristas)')

# Un step
edge_0 = env.all_edges[0]
obs2, reward, terminated, truncated, info = env.step(0)
print(f'\nStep 0 — flip arista {edge_0}:')
print(f'  scalar reward : {reward:.6f}')
print(f'  r_priv        : {info["reward_vec"][0]:.6f}')
print(f'  r_util        : {info["reward_vec"][1]:.6f}')
print(f'  terminated    : {terminated}')

# Rollout aleatorio (linea de base inferior)
obs_r, _ = env.reset()
cum_priv = cum_util = 0.0
for _ in range(env.T):
    a = env.action_space.sample()
    obs_r, _, done, _, info_r = env.step(a)
    cum_priv += info_r['reward_vec'][0]
    cum_util += info_r['reward_vec'][1]
    if done:
        break

print(f'\nRollout aleatorio (T={env.T})  [linea de base inferior para MOPPO]:')
print(f'  R_priv acumulado : {cum_priv:.4f}')
print(f'  R_util acumulado : {cum_util:.4f}')
```

## 4. Test `MOPPOActorCritic`

Verifica: parametros, forward pass, `get_action` (estocastico y deterministico), entropia.

```python
import torch.nn.functional as F
from graph_anon_morl.models import MOPPOActorCritic

obs_dim   = env.observation_space.shape[0]
n_actions = env.n_actions

model = MOPPOActorCritic(obs_dim, n_actions, hidden_dim=256)
n_params = sum(p.numel() for p in model.parameters())

print('MOPPOActorCritic:')
print(f'  obs_dim    : {obs_dim}')
print(f'  n_actions  : {n_actions}')
print(f'  parametros : {n_params:,}')

# Forward pass
obs_test, _ = env.reset()
obs_t = torch.FloatTensor(obs_test).unsqueeze(0)
logits, value = model(obs_t)
print(f'\nForward pass:')
print(f'  logits.shape : {logits.shape}')
print(f'  value        : {value.item():.4f}')

# get_action estocastico
action, logp, entropy, _ = model.get_action(obs_t, deterministic=False)
max_entropy = np.log(n_actions)
print(f'\nget_action (estocastico):')
print(f'  action       : {action.item()}')
print(f'  log_prob     : {logp.item():.4f}')
print(f'  entropy      : {entropy.item():.4f}')
print(f'  entropy_norm : {entropy.item() / max_entropy:.4f}  (1.0 = uniforme)')

# get_action deterministico
action_d, _, _, _ = model.get_action(obs_t, deterministic=True)
print(f'\nget_action (deterministico)  : action={action_d.item()} (argmax logits)')

# Distribucion inicial (recien inicializado ~ uniforme)
probs = F.softmax(logits, dim=-1).detach().numpy().flatten()
print(f'\nDistribucion de probs (inicializacion):')
print(f'  max prob : {probs.max():.6f}  (uniforme seria {1/n_actions:.6f})')
print(f'  min prob : {probs.min():.6f}')
print(f'  std prob : {probs.std():.6f}')
```

## 5. Mini Entrenamiento (10 iteraciones)

Smoke test del loop MOPPO completo. Para resultados del paper usar `train_main.py --graph facebook --n_nodes 100 --n_iter 500`.

```python
from train_main import collect_rollout, compute_gae, ppo_update

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Device: {device}')

torch.manual_seed(42)
np.random.seed(42)

N_ITER  = 10
N_STEPS = 256

model_train = MOPPOActorCritic(obs_dim, n_actions, hidden_dim=128).to(device)
optimizer   = optim.Adam(model_train.parameters(), lr=3e-4)
env_train   = GraphAnonEnv(G_fb, k=2, T=50)

metrics = []
print(f'\nEntrenando {N_ITER} iters (n_steps={N_STEPS}) sobre ego-Facebook n=100...')
for i in range(N_ITER):
    obs_b, acts_b, logps_b, vals_b, rews_b, dones_b = collect_rollout(
        env_train, model_train, N_STEPS, device
    )
    advs, rets = compute_gae(rews_b, vals_b, dones_b)
    loss = ppo_update(model_train, optimizer, obs_b, acts_b, logps_b, rets, advs)
    mean_r = rews_b.mean().item()
    metrics.append({'iter': i + 1, 'loss': loss, 'mean_reward': mean_r})
    print(f'  [{i+1:2d}/{N_ITER}]  loss={loss:.5f}   mean_r={mean_r:.5f}')

df_m = pd.DataFrame(metrics)
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(df_m['iter'], df_m['loss'], marker='o', markersize=4, color='steelblue')
axes[0].set_xlabel('Iteracion')
axes[0].set_ylabel('Loss (PPO)')
axes[0].set_title('Training Loss')

axes[1].plot(df_m['iter'], df_m['mean_reward'], marker='o', markersize=4, color='#43A047')
axes[1].set_xlabel('Iteracion')
axes[1].set_ylabel('Reward medio escalarizado')
axes[1].set_title('Mean Reward')

plt.suptitle(f'MOPPO — ego-Facebook n=100  ({N_ITER} iters, n_steps={N_STEPS})')
plt.tight_layout()
plt.show()
```

## 6. Pareto Frontier Sweep

Barre alpha en [0,1] con 11 puntos y evalua el frente de Pareto (privacidad vs utilidad).

```python
from graph_anon_morl.evaluation import evaluate_policy, compute_pareto_front, hypervolume_indicator

env_eval = GraphAnonEnv(G_fb, k=2, T=50)
model_train.eval()

results    = evaluate_policy(env_eval, model_train, n_weights=11, n_episodes=5, device=str(device))
reward_vecs = [(r[1], r[2]) for r in results]
pareto_idx  = compute_pareto_front(reward_vecs)
pareto_pts  = [reward_vecs[i] for i in pareto_idx]
hv = hypervolume_indicator(pareto_pts)

print(f'Pareto front size : {len(pareto_pts)}')
print(f'Hypervolume       : {hv:.6f}')
print()
print(f'{"alpha":>8s}  {"r_priv":>10s}  {"r_util":>10s}  {"pareto":>7s}')
print('-' * 42)
for alpha, rp, ru in results:
    flag = '*' if (rp, ru) in pareto_pts else ' '
    print(f'{alpha:8.2f}  {rp:10.5f}  {ru:10.5f}  {flag:>7s}')

alphas  = [r[0] for r in results]
r_privs = [r[1] for r in results]
r_utils = [r[2] for r in results]

fig, ax = plt.subplots(figsize=(8, 5))
sc = ax.scatter(r_privs, r_utils, c=alphas, cmap='viridis', s=110,
                zorder=3, edgecolors='white', linewidths=0.5)
if pareto_pts:
    px, py = zip(*pareto_pts)
    ax.scatter(px, py, marker='*', s=280, color='red',
               label=f'Pareto front  (n={len(pareto_pts)})', zorder=4)
plt.colorbar(sc, ax=ax, label='alpha (peso privacidad)')
ax.set_xlabel('r_priv  (k-anonimato, mayor = mas privado)')
ax.set_ylabel('r_util  (clustering coeff, mayor = mas util)')
ax.set_title(f'MOPPO Pareto Frontier — ego-Facebook (n=100)\nHypervolume = {hv:.4f}')
ax.legend()
plt.tight_layout()
plt.show()
```

## 7. Fairness Audit

Analiza si el costo de anonimizacion es desproporcionado para nodos perifericos (Q1) vs hubs (Q4). `disparity > 1` indica sesgo contra nodos de bajo grado — hallazgo clave del paper.

```python
from graph_anon_morl.audit import degree_quartile_audit, fairness_disparity_index

# Episodio deterministico con alpha=0.5
obs_a, _ = env_eval.reset(weight=[0.5, 0.5])
done = False
trajectory = []

with torch.no_grad():
    while not done:
        obs_t = torch.FloatTensor(obs_a).unsqueeze(0).to(device)
        action, _, _, _ = model_train.get_action(obs_t, deterministic=True)
        obs_a, _, terminated, truncated, info = env_eval.step(action.item())
        trajectory.append(info['reward_vec'].tolist())
        done = terminated or truncated

audit_stats = degree_quartile_audit(env_eval.G0, env_eval.G)
disparity   = fairness_disparity_index(audit_stats)

print('Fairness Audit — Perdida de aristas por cuartil de grado:')
print(f'{"Grupo":20s} {"n_nodos":>8s} {"deg_G0":>8s} {"edge_loss":>10s} {"norm_loss":>10s}')
print('-' * 60)
for group, s in audit_stats.items():
    n_nd = s['n_nodes']
    adg  = s['avg_degree_G0']
    ael  = s['avg_edge_loss']
    nl   = s['normalized_loss']
    print(f'{group:20s} {n_nd:8d} {adg:8.2f} {ael:10.3f} {nl:10.4f}')
print()
interp = 'nodos perifericos soportan MAS costo' if disparity > 1 else 'distribucion equitativa'
print(f'Disparity Q1/Q4 : {disparity:.4f}  ->  {interp}')

# Visualizacion
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

groups      = list(audit_stats.keys())
norm_losses = [audit_stats[g]['normalized_loss'] for g in groups]
bar_cols    = ['#2196F3', '#4CAF50', '#F44336']
bars = axes[0].bar(groups, norm_losses, color=bar_cols, edgecolor='white')
axes[0].set_ylabel('Perdida de aristas normalizada')
axes[0].set_title(f'Costo de anonimizacion por cuartil\nDisparity = {disparity:.3f}')
axes[0].tick_params(axis='x', rotation=15)
for bar, val in zip(bars, norm_losses):
    axes[0].text(bar.get_x() + bar.get_width() / 2.0,
                 bar.get_height() + 0.002,
                 f'{val:.3f}', ha='center', va='bottom', fontsize=9)

traj_arr = np.array(trajectory)
axes[1].plot(traj_arr[:, 0], label='r_priv', color='#E53935', linewidth=1.5)
axes[1].plot(traj_arr[:, 1], label='r_util', color='#1E88E5', linewidth=1.5)
axes[1].set_xlabel('Paso')
axes[1].set_ylabel('Reward')
axes[1].set_title('Trayectoria de rewards  (alpha=0.5)')
axes[1].legend()

plt.tight_layout()
plt.show()
```

## 8. TREX Trajectory Clustering

Agrupa 30 trayectorias por perfil acumulado [R_priv, R_util] (estilo TREX). Permite identificar si el agente adopta distintos modos de comportamiento segun alpha.

```python
from graph_anon_morl.audit import trex_trajectory_cluster

N_TRAJ = 30
rng    = np.random.default_rng(seed=42)
model_train.eval()
trajectories = []

with torch.no_grad():
    for _ in range(N_TRAJ):
        alpha    = float(rng.uniform(0, 1))
        obs_tr, _ = env_eval.reset(weight=[alpha, 1.0 - alpha])
        done = False
        traj = []
        while not done:
            obs_tt = torch.FloatTensor(obs_tr).unsqueeze(0).to(device)
            action, _, _, _ = model_train.get_action(obs_tt, deterministic=False)
            obs_tr, _, terminated, truncated, info = env_eval.step(action.item())
            traj.append(info['reward_vec'].tolist())
            done = terminated or truncated
        trajectories.append(traj)

labels, features, centers = trex_trajectory_cluster(trajectories, n_clusters=3)

print('TREX Clustering — Clusters de trayectorias:')
print(f'{"Cluster":>8s}  {"R_priv_sum":>12s}  {"R_util_sum":>12s}  {"n_traj":>7s}')
print('-' * 46)
for i, c in enumerate(centers):
    n_traj_i = int((labels == i).sum())
    print(f'{i:8d}  {c[0]:12.3f}  {c[1]:12.3f}  {n_traj_i:7d}')

fig, ax = plt.subplots(figsize=(7, 5))
sc = ax.scatter(features[:, 0], features[:, 1], c=labels,
                cmap='tab10', s=80, zorder=3, edgecolors='white', linewidths=0.4)
ax.scatter(centers[:, 0], centers[:, 1], marker='X', s=220,
           color='black', label='Centroides', zorder=4)
ax.set_xlabel('R_priv acumulado')
ax.set_ylabel('R_util acumulado')
ax.set_title(f'TREX Trajectory Clustering  (n={N_TRAJ} trayectorias, k=3)')
ax.legend()
plt.colorbar(sc, ax=ax, label='Cluster')
plt.tight_layout()
plt.show()
```
